```python
import pandas as pd

df_bridge = pd.read_csv("../data/processed/geography_bridge.csv",
    dtype={
        "geo_id": str,
        "redfin_region_id": str,
        "zillow_region_id": str
    }
)
df_bridge.head()
```

```python
df_redfin = pd.read_csv("../data/processed/redfin_monthly_housing.csv",dtype={"redfin_region_id": str})
```

```python
df_redfin_test = df_redfin.merge(df_bridge[["geo_id", "redfin_region_id"]],
    on="redfin_region_id",
    how="inner",
    validate="many_to_one"
)
```

```python
df_zillow = pd.read_csv("../data/processed/zillow_monthly_rent.csv",dtype={"zillow_region_id": str})

df_zillow_test = df_zillow.merge(df_bridge[["geo_id", "zillow_region_id"]],
    on="zillow_region_id",
    how="inner",
    validate="many_to_one"
)
```

```python
df_census = pd.read_csv("../data/processed/census_geography_income_2024.csv",dtype={"geo_id": str})

df_redfin["redfin_region_id"] = (df_redfin["redfin_region_id"].astype(str))

redfin_geo = (df_redfin[["redfin_region_id", "region_name"]].drop_duplicates())

census_geo = (df_census[["geo_id", "geo_name", "geography_type"]].drop_duplicates())

redfin_census = redfin_geo.merge(
    census_geo,
    left_on="redfin_region_id",
    right_on="geo_id",
    how="inner",
    validate="one_to_one"
)

print("Redfin/Census compatible geographies:",len(redfin_census))
```

```python
redfin_census_ids = set(redfin_census["redfin_region_id"])

final_bridge_ids = set(df_bridge["redfin_region_id"])

not_in_final_bridge = (redfin_census_ids - final_bridge_ids)

print("Redfin/Census markets not in final bridge:",len(not_in_final_bridge))
```

```python
redfin_census[
    redfin_census["redfin_region_id"].isin(not_in_final_bridge)
][
    [
        "redfin_region_id",
        "region_name",
        "geo_name",
        "geography_type"
    ]
].sort_values("region_name")
```
